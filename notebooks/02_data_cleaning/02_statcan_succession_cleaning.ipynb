# Data Cleaning: Statistics Canada Succession Tables

**Project:** Niagara Succession Signal  
**Stage:** 02 - Data Cleaning  
**Scope:** Week 4 - Succession Risk Signal Data Cleaning  

## Objectives
1. Ingest raw Statistics Canada quarterly/annual succession survey tables:
   - Ù01196.csv\: Intentions for businesses to sell or transfer ownership within 10 years
   - Ù01197.csv\: Expectations related to selling or transferring the business
   - Ù01198.csv\: Considerations related to the sale or transfer of the business
   - Ù01199.csv\: Scenarios if business is unable to find a suitable buyer or successor
2. Standardize geographic scope to Ontario (\DGUID == '2021A000235'\).
3. Clean categorical text, audit \STATUS\ suppression codes (\F\, \..\), verify coordinate mapping, and strip unpopulated fields (\SYMBOL\).
4. Export processed provincial slices to \data/interim/\ for downstream feature engineering.

```python
import os
import pandas as pd
import numpy as np

# Configure project paths
RAW_DIR = os.path.join('..', '..', 'data', 'raw', 'statcan', 'succession')
INTERIM_DIR = os.path.join('..', '..', 'data', 'interim')
os.makedirs(INTERIM_DIR, exist_ok=True)

ONTARIO_DGUID = '2021A000235'
```

## 1. Table 33-10-1196: Business Succession Intentions

```python
df_1196_raw = pd.read_csv(os.path.join(RAW_DIR, '33101196.csv'))
print(f"Raw Table 1196 rows: {len(df_1196_raw):,}, columns: {len(df_1196_raw.columns)}")

# Filter for Ontario
df_1196_ont = df_1196_raw[df_1196_raw['DGUID'] == ONTARIO_DGUID].copy()
if 'SYMBOL' in df_1196_ont.columns and df_1196_ont['SYMBOL'].isna().all():
    df_1196_ont.drop(columns=['SYMBOL'], inplace=True)

# Strip string padding
str_cols = df_1196_ont.select_dtypes(include=['object', 'string']).columns
for c in str_cols:
    df_1196_ont[c] = df_1196_ont[c].str.strip()

print(f"Ontario Table 1196 rows: {len(df_1196_ont):,}")
print("STATUS distribution:")
print(df_1196_ont['STATUS'].value_counts(dropna=False))

df_1196_ont.to_csv(os.path.join(INTERIM_DIR, 'statcan_33101196_ontario_cleaned.csv'), index=False)
```
